## Kata 1: 

Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def frecuencia_letras(texto):
    frecuencias = {}

    # Recorremos cada carácter del texto
    for letra in texto:
        # Ignoramos los espacios
        if letra != " ":
            frecuencias[letra] = frecuencias.get(letra, 0) + 1

    return frecuencias


texto_ejemplo = "hola ke ase, programa o ke ase"

print(frecuencia_letras(texto_ejemplo))
```

## Kata 5:

Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe devolver una tupla que contenga la media y el estado.

```python
def calcular_media_estado(notas, nota_aprobado=5):
    # Media simple
    media = sum(notas) / len(notas)
    # Comprobación
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"

    return media, estado

# Probamos dos listas de notas
resultado = calcular_media_estado([6, 7, 4, 8])
resultado_dos = calcular_media_estado([3, 2, 4, 5])

print(resultado)
print(resultado_dos)
```

## Kata 6: 

Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(numero):
    # Controlamos los negativos
    if numero < 0:
        raise ValueError("El número debe ser mayor o igual que cero.")

    # Caso base. Pensaba que el factorial de 0 era 0, pero resulta que es uno, si no peta, claro
    if numero == 0 or numero == 1:
        return 1

    # La función se llama a sí misma para resolver de manera recursiva
    return numero * factorial(numero - 1)


print(factorial(5))
```

Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.


```python
# Excepción personalizada
class ListaVaciaError(Exception):
    pass

# Función que recibe una lista de números
def calcular_promedio_seguro(numeros):
    # Si está vacía, lanzamos la excepción
    if not numeros:
        raise ListaVaciaError("No se puede calcular el promedio de una lista vacía.")
    # Sumamos todos los números y dividimos entre la cantidad de elementos de la lista
    promedio = sum(numeros) / len(numeros)

    return promedio


# Intentamos ejecutar la función
try:
    numeros = [5, 7, 9, 6]

    resultado = calcular_promedio_seguro(numeros)

    print(f"El promedio es: {resultado}")

# Si la lista está vacía, recogemos la excepción
except ListaVaciaError as error:
    print(f"Error: {error}")
```

## Kata 13:

Genera una función que, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas. Las letras no pueden estar repetidas. Usa la función map().


```python
def caracteres_mayus_minus(caracteres):

    # Con dict.fromkeys() eliminamos los caracteres repetidos manteniendo el orden en el que aparecen
    caracteres_unicos = list(dict.fromkeys(caracteres))

    # Función que recibe un caracter y devuelve una tupla con su versión en mayúscula
    # y su versión en minúscula
    def convertir_caracter(caracter):
        return (caracter.upper(), caracter.lower())

    # Map aplica convertir_caracter a cada caracter único y convertimos el resultado en una lista
    resultado = list(map(convertir_caracter, caracteres_unicos))

    # Convertimos el resultado de map en una lista
    return resultado

caracteres = "Python Verbenero"

print(caracteres_mayus_minus(caracteres))
```

## Kata 14: 

Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().


```python
def palabras_por_letra(palabras, letra):

    # Función que comprueba si una palabra empieza por la letra indicada
    def empieza_por_letra(palabra):

        # Pasamos ambos valores a minúsculas 
        return palabra.lower().startswith(letra.lower())

    # Filter aplica empieza_por_letra a cada palabra y se queda solo las que devuelven True
    # Convertimos el resultado en una lista
    resultado = list(filter(empieza_por_letra, palabras))

    return resultado

palabras = ["Casa", "coche", "Python", "camino", "Java"]

print(palabras_por_letra(palabras, "c"))
```

## Kata 16:

Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las palabras que sean más largas que n. Usa la función filter().


```python
# Función que recibe un texto y un número entero
def palabras_largas(texto, n):

    # split() divide el texto y crea una lista de palabras
    palabras = texto.split()

    # Devuelve True cuando la palabra tiene más caracteres que el valor de n
    def supera_longitud(palabra):
        return len(palabra) > n

    # Filter para quedarnos con las palabras que devuelven True. Convertimos a lista
    resultado_filter = list(filter(supera_longitud, palabras))

    return resultado_filter

texto = "Me salen las katas por las orejas"

# Mostramos las palabras que tienen más de 4 caracteres
print(palabras_largas(texto, 4))
```

## Kata 17:

Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().


```python
# Importamos reduce 
from functools import reduce

# Función que recibe:
# + acumulado: el texto que se ha ido formando
# + digito: el siguiente número de la lista
def unir_digitos(acumulado, digito):

    # Pasamos el dígito a texto con str()
    # y lo añadimos al texto acumulado
    return acumulado + str(digito)

# Función que recibe una lista de dígitos
def digitos_a_numero(digitos):

    # Reduce hace unir_digitos sobre todos los elementos
    # Empezamos con una cadena vacía
    numero_texto = reduce(unir_digitos, digitos, "")

    # Pasamos el texto final a entero
    return int(numero_texto)

digitos = [5, 7, 2]

print(digitos_a_numero(digitos))
```

## Kata 28:

Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.


```python
#Función que recibe una lista de elementos y comprueba si están repes
def primer_duplicado(elementos):

    # Lista vacía para guardar los elementos que ya hemos recorrido
    elementos_vistos = []

    # Recorremos la lista original
    for elemento in elementos:

        # Si el elemento ya está en la lista, significa que está repe y es el primero que buscamos
        if elemento in elementos_vistos:
            return elemento

        # Si todavía no estaba, lo guardamos
        elementos_vistos.append(elemento)

    # Si no encontramos ningún repetido
    return None


lista = [3, 5, 1, 4, 5, 3, 1]

print(primer_duplicado(lista))
```

## Kata 29:

Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.


```python
# Funcion que recoge una variable, la convierte a texto 
# y la enmascara, dejando los últimos 4 sin enmascarar
def enmascarar(variable):
    # Pasamos a texto
    texto = str(variable)

    # Si tiene cuatro caracteres o menos, no hay nada que ocultar
    if len(texto) <= 4:
        return texto

    # Calculamos cuántos caracteres debemos ocultar
    cantidad_oculta = len(texto) - 4

    # Creamos una cadena de # igual de larga que la parte oculta
    parte_oculta = "#" * cantidad_oculta

    # Obtenemos los últimos cuatro caracteres
    ultimos_cuatro = texto[-4:]

    # Unimos la parte oculta con los últimos cuatro caracteres
    return parte_oculta + ultimos_cuatro


print(enmascarar("1234567890123456"))
```

## Kata 30:

Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.


```python
# Función que recibe 2 palabras para comprobar si son anagramas
def son_anagramas(palabra1, palabra2):
    # Pasamos todo a minúsculas, porque sorted si no se ralla 
    # y pone las mayúsculas primero al ordenarlas
    minusculas1 = palabra1.lower()
    minusculas2 = palabra2.lower()

    # Sorted descompone las palabras y ordena las letras para poder compararlas
    # Si son iguales, nos lanzará true
    return sorted(minusculas1) == sorted(minusculas2)


print(son_anagramas("Roma", "Amor"))
print(son_anagramas("Patatas", "Fritas"))
```

## Kata 31:

Crea una función que solicite al usuario ingresar una lista de nombres y luego un nombre para buscar en esa lista. Si el nombre está en la lista, imprime un mensaje indicando que fue encontrado; de lo contrario, lanza una excepción.

```python
def buscar_nombre_usuario():

    try:
        # Pedimos varios nombres separados por comas
        entrada = input("Introduce nombres separados por comas: ")

        # Separamos los nombres usando la coma
        nombres_separados = entrada.split(",")

        # Creamos una lista vacía para guardar los nombres sin espacios sobrantes
        nombres = []

        # Recorremos los nombres introducidos
        for nombre in nombres_separados:
            # strip() para limpiar los espacios al principio y al final y los guardamos en la lista
            nombres.append(nombre.strip())

        # Pedimos el nombre que queremos buscar, con strip() para limpiar espacios
        nombre_buscado = input(
            "Introduce el nombre que quieres buscar: "
        ).strip()

        # Si el nombre no aparece en la lista, lanzamos excepción
        if nombre_buscado not in nombres:
            raise ValueError(
                f"{nombre_buscado} no se encuentra en la lista."
            )

        # Lanzamos mensaje si el nombre existe
        print(f"{nombre_buscado} fue encontrado.")

    except ValueError as error:
        # Capturamos el error y enseñamos mensaje
        print(f"Error: {error}")


buscar_nombre_usuario()
```

## Kata 32:

Crea una función que tome un nombre completo y una lista de empleados, busque el nombre en la lista y devuelva el puesto del empleado si se encuentra; de lo contrario, devuelve un mensaje indicando que la persona no trabaja aquí.

```python
# Función que recibe el nombre que queremos buscar y una lista con los datos de los empleados
def buscar_puesto(nombre_completo, empleados):

    # Recorremos uno por uno todos los empleados
    for empleado in empleados:
        # Comparamos el nombre buscado con el nombre del empleado y usamos lower()
        if empleado["nombre"].lower() == nombre_completo.lower():
            # Si encontramos una coincidencia devolvemos el puesto del empleado
            return empleado["puesto"]

    # Si termina el bucle y no aparece la persona devolvemos este mensaje
    return "Esta persona no trabaja aquí."


# Creamos una lista de empleados con diccionarios
empleados = [
    {"nombre": "Ana García", "puesto": "Desarrolladora"},
    {"nombre": "Luis Pérez", "puesto": "Diseñador"},
    {"nombre": "Marta López", "puesto": "Administradora"}
]

print(buscar_puesto("Luis Pérez", empleados))
print(buscar_puesto("Carlos Ruiz", empleados))
```

```python
# Clase Arbol
class Arbol:

    # Método para inicializar el arbol
    def __init__(self):

        # El tronco comienza con longitud 1
        self.tronco = 1

        # El árbol comienza sin ramas
        self.ramas = []

    # Aumenta la longitud del tronco en una unidad
    def crecer_tronco(self):
        self.tronco += 1

    # Añade una nueva rama de longitud 1
    def nueva_rama(self):
        self.ramas.append(1)

    # Aumenta en una unidad todas las ramas
    def crecer_ramas(self):

        # Recorremos las posiciones de la lista
        for posicion in range(len(self.ramas)):
            # Y las hacemos crecer 1
            self.ramas[posicion] += 1

    # Elimina una rama según su posición
    def quitar_rama(self, posicion):

        # Restamos uno para empezar en la posición 0
        indice = posicion - 1

        # Comprobamos que la posición exista
        if indice < 0 or indice >= len(self.ramas):
            raise IndexError("La posición de la rama no existe.")

        # Eliminamos la rama
        self.ramas.pop(indice)

    # Devuelve la información del árbol
    def info_arbol(self):
        return {
            "longitud_tronco": self.tronco,
            "numero_ramas": len(self.ramas),
            "longitud_ramas": self.ramas
        }


# Caso de uso
arbol = Arbol()
arbol.crecer_tronco()
arbol.nueva_rama()
arbol.crecer_ramas()
arbol.nueva_rama()
arbol.nueva_rama()
arbol.quitar_rama(2)

print(arbol.info_arbol())
```

## Kata 35: 

Crea la clase `UsuarioBanco`

- Representa a un usuario de un banco con su nombre, saldo y un indicador que señale si tiene o no cuenta corriente.
- Métodos disponibles: `retirar_dinero`, `transferir_dinero`, `agregar_dinero`.

### Código a seguir

1. Inicializar un usuario con:
   - Nombre.
   - Saldo.
   - Un indicador (`True` o `False`) de si tiene cuenta corriente.
2. Implementar el método `retirar_dinero` para sustraer dinero del saldo, lanzando un error si la operación no es posible.
3. Implementar el método `transferir_dinero` para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
4. Implementar el método `agregar_dinero` para aumentar el saldo del usuario.

### Caso de uso

1. Crear dos usuarios:
   - `"Alicia"`, con un saldo inicial de 100 y cuenta corriente.
   - `"Bob"`, con un saldo inicial de 50 y cuenta corriente.
2. Agregar 20 unidades al saldo de Bob.
3. Transferir 80 unidades de Bob a Alicia.
4. Retirar 50 unidades del saldo de Alicia.

```python
# Clase UsuarioBanco
class UsuarioBanco:

    # Método para inicializar un nuevo usuario
    def __init__(self, nombre, saldo, cuenta_corriente):

        # Guardamos el nombre del usuario
        self.nombre = nombre

        # Guardamos el saldo inicial del usuario
        self.saldo = saldo

        # Guardamos si el usuario tiene cuenta corriente (true o false)
        self.cuenta_corriente = cuenta_corriente


    # Método para retirar dinero, le pasamos la cantidad que queremos retirar
    def retirar_dinero(self, cantidad):

        # Controlamos que la cantidad sea mayor que cero
        if cantidad <= 0:

            # Lanzamos un error si la cantidad no es válida
            raise ValueError(
                "La cantidad debe ser mayor que cero."
            )

        # Comprobamos si el usuario tiene suficiente dinero
        if cantidad > self.saldo:

            # Lanzamos un error si intenta retirar más dinero del que tiene 
            raise ValueError("Saldo insuficiente.")

        # Si todo va bien, restamos la cantidad al saldo del usuario
        self.saldo -= cantidad

    # Método para añadir dinero al saldo
    def agregar_dinero(self, cantidad):

        # Comprobamos que la cantidad sea mayor que cero
        if cantidad <= 0:

            # Lanzamos un error si la cantidad no es válida
            raise ValueError(
                "La cantidad debe ser mayor que cero."
            )

        # Si la cantidad es válida, la sumamos al saldo del usuario
        self.saldo += cantidad
    
    # Método para transferir dinero. "Destinatario" es el usuario que recibirá el dinero
    # y "cantidad" es el dinero que queremos transferir
    def transferir_dinero(self, destinatario, cantidad):

        # Comprobamos que el usuario que envía el dinero y el que lo recibe
        # tengan  cuenta corriente
        if (
            not self.cuenta_corriente
            or not destinatario.cuenta_corriente
        ):
            # Si alguno no tiene, lanzamos error
            raise ValueError(
                "Ambos usuarios deben tener cuenta corriente."
            )

        # Sacamos el dinero del usuario que envía el dinero llamando a la función retirar_dinero()
        self.retirar_dinero(cantidad)

        # Se lo ingresamos al destinatario con agregar_dinero()
        destinatario.agregar_dinero(cantidad)


# CASO DE USO

alicia = UsuarioBanco("Alicia", 100, True)

bob = UsuarioBanco("Bob", 50, True)

# Añadimos 20 al saldo de Bob
bob.agregar_dinero(20)

# Utilizamos try porque la transferencia puede provocar un ValueError
try:
    # Bob intenta transferirle 80 a Alicia
    bob.transferir_dinero(alicia, 80)

# Si se produce un ValueError guardamos el mensaje del error en la variable error
except ValueError as error:

    # Mostramos un mensaje
    print(
        f"No se pudo realizar la transferencia: {error}"
    )

# Alicia retira 50 de su cuenta
alicia.retirar_dinero(50)

# Mostramos el saldo final de Alicia
print(f"Saldo de Alicia: {alicia.saldo}")

# Mostramos el saldo final de Bob
print(f"Saldo de Bob: {bob.saldo}")
```

## Kata 36: 

Crea una función llamada `procesar_texto`

- Procesa un texto según la opción especificada: `contar_palabras`, `reemplazar_palabras` o `eliminar_palabra`.

### Código a seguir

1. Crear una función `contar_palabras` que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
2. Crear una función `reemplazar_palabras` para sustituir una `palabra_original` por una `palabra_nueva` en el texto y devolver el texto modificado.
3. Crear una función `eliminar_palabra` que elimine una palabra del texto y devuelva el texto sin ella.
4. Crear la función `procesar_texto` que reciba un texto, una opción (`"contar"`, `"reemplazar"`, `"eliminar"`) y un número variable de argumentos según la opción elegida.

### Caso de uso

- Verificar el funcionamiento completo de `procesar_texto`.

```python
# Cuenta cuántas veces aparece cada palabra
def contar_palabras(texto):
    frecuencias = {}

    # Separamos el texto en palabras
    for palabra in texto.split():

        # Pasamos a minúsculas
        palabra_minus = palabra.lower()

        # Si ya existe, aumentamos su contador
        if palabra_minus in frecuencias:
            frecuencias[palabra_minus] += 1

        # Si no existe, la añadimos con valor 1
        else:
            frecuencias[palabra_minus] = 1

    return frecuencias


# Sustituye una palabra por otra
def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    return texto.replace(palabra_original, palabra_nueva)


# Elimina una palabra del texto
def eliminar_palabra(texto, palabra_eliminar):
    palabras_restantes = []

    # Recorremos las palabras del texto
    for palabra in texto.split():

        # Solo guardamos las que no queremos eliminar
        if palabra != palabra_eliminar:
            palabras_restantes.append(palabra)

    # Volvemos a unir las palabras con espacios
    return " ".join(palabras_restantes)


# Función principal que llama a las demás
def procesar_texto(texto, opcion, *args):

    if opcion == "contar":
        return contar_palabras(texto)

    elif opcion == "reemplazar":

        return reemplazar_palabras(texto, args[0], args[1])

    elif opcion == "eliminar":

        return eliminar_palabra(texto, args[0])

    else:
        raise ValueError(
            "La opción debe ser contar, reemplazar o eliminar."
        )


texto = "Empiezo a tener pesadillas con Python"

# Probamos las tres opciones
print(procesar_texto(texto, "contar"))
print(procesar_texto(texto, "reemplazar", "Python", "JS"))
print(procesar_texto(texto, "eliminar", "pesadillas"))
```

## Kata 39:

Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).


```python
# Función que recibe una figura (nombre en texto) y datos (tupla con las medidas)
def calcular_area(figura, datos):

    # Convertimos el nombre de la figura a minúsculas
    figura = figura.lower()

    # Comprobamos si la figura es un rectángulo
    if figura == "rectangulo":

        # Un rectángulo necesita base y altura
        if len(datos) != 2:

            # Si no recibimos exactamente dos valores lanzamos un error
            raise ValueError(
                "El rectángulo necesita base y altura."
            )

        # Sacamos los dos valores de la tupla, el primer valor se guarda en base
        # y el segundo en altura
        base, altura = datos

        # Calculamos el área (base * altura) y la devolvemos
        area = base * altura

        return area

    # Si no es un rectángulo comprobamos si es un círculo
    elif figura == "circulo":

        # Un círculo el radio
        if len(datos) != 1:

            # Si recibimos más o menos de un valor lanzamos un error
            raise ValueError(
                "El círculo necesita el radio."
            )

        # Obtenemos el primer (y único) valor de la tupla
        radio = datos[0]

        # Usamos 3.1416 como valor aproximado de pi
        pi = 3.1416

        # Calculamos el área (pi * radio al cuadrado)
        area = pi * radio ** 2

        # Devolvemos el resultado
        return area

    # Si no es un rectángulo ni un círculo comprobamos si es un triángulo
    elif figura == "triangulo":

        # Necesita dos datos, base y altura
        if len(datos) != 2:

            # Si no recibimos dos valores lanzamos un error
            raise ValueError(
                "El triángulo necesita base y altura."
            )

        # Extraemos la base y la altura de la tupla
        base, altura = datos

        # Calculamos el área (base * altura / 2)
        area = base * altura / 2

        # Devolvemos el resultado
        return area

    # Si la figura no coincide con ninguna de las opciones permitidas
    else:

        # Lanzamos un error indicando cuáles son las figuras válidas
        raise ValueError(
            "La figura debe ser rectangulo, circulo o triangulo."
        )
        
# Calculamos el área de un rectángulo de base 5 y altura 3
print(calcular_area("rectangulo", (5, 3)))

# Calculamos el área de un círculo de radio 4. Con coma es para crear una tupla de un elemento
print(calcular_area("circulo", (4,)))

# Calculamos el área de un triángulo de base 6 y altura 2
print(calcular_area("triangulo", (6, 2)))
```

## Kata 40: 

Descuento en una tienda en línea

Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en línea después de aplicar un descuento.

### Código a seguir

1. Solicitar al usuario el precio original de un artículo.
2. Preguntar si tiene un cupón de descuento (`sí` o `no`).
3. Si la respuesta es sí, solicitar el valor del cupón de descuento.
4. Aplicar el descuento al precio original, siempre que el valor del cupón sea mayor que cero.
5. Mostrar el precio final de la compra, considerando o no el descuento.
6. Usar estructuras de control de flujo (`if`, `elif`, `else`).

```python
# Función que calcula el precio final de la compra
def calcular_precio_final(precio_original, tiene_cupon, valor_cupon=0):

    # Comprobamos que el precio sea válido
    if precio_original < 0:
        raise ValueError("El precio no puede ser negativo.")

    # Normalizamos la respuesta
    respuesta = tiene_cupon.lower()

    # Si tiene cupón
    if respuesta == "sí" or respuesta == "si":

        # Comprobamos que el cupón sea válido
        if valor_cupon <= 0:
            raise ValueError(
                "El valor del cupón debe ser mayor que cero."
            )

        # Aplicamos el descuento
        precio_final = precio_original - ((precio_original * valor_cupon) / 100)

        # Evitamos precios negativos
        if precio_final < 0:
            precio_final = 0

        return precio_final

    # Si no tiene cupón
    elif respuesta == "no":
        return precio_original

    # Si escribe otra cosa
    else:
        raise ValueError(
            "La respuesta debe ser sí o no."
        )

# Creamos una función para pedir los datos al usuario
def programa_descuento():

    try:
        # Pedimos el precio original. Usamos float() porque el input devuelve texto
        precio = float(
            input("Introduce el precio original: ")
        )

        # Preguntamos si tiene cupón
        respuesta = input(
            "¿Tienes un cupón de descuento? Sí o no: "
        )

        # Inicialmente el cupón vale 0
        valor_cupon = 0

        # Solo pedimos el valor si responde que sí
        if respuesta.lower() == "sí" or respuesta.lower() == "si":
            valor_cupon = float(
                input("Introduce el valor del cupón (% de descuento): ")
            )

        # Pasamos todos los datos a la función que nos calcula el precio final
        precio_final = calcular_precio_final(
            precio,
            respuesta,
            valor_cupon
        )

        print(
            f"El precio final es: {precio_final:.2f} €"
        )

    except ValueError as error:
        print(f"Error: {error}")


programa_descuento()
```
